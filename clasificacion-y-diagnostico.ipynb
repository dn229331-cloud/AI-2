# clasificacion y diagnostico

Introducción a MNIST

Conjunto de datos clásico compuesto por miles de imágenes de dígitos escritos a mano (del 0 al 9), utilizado universalmente como benchmark para entrenar y evaluar algoritmos de visión por computadora y clasificación.

Clasificador Binario

Modelo de aprendizaje supervisado optimizado para responder una sola pregunta de tipo sí/no o de dos clases mutuamente excluyentes (por ejemplo: "¿Esta imagen es el dígito 5 o no lo es?").

Matriz de Confusión

Tabla de contingencia que desglosa las predicciones de un modelo frente a las etiquetas reales del dataset, dividiendo los resultados en cuatro cuadrantes: Verdaderos Positivos (VP), Falsos Positivos (FP), Verdaderos Negativos (VN) y Falsos Negativos (FN).

Permite derivar métricas críticas de rendimiento como el Recall (la proporción de casos positivos reales que el modelo logró identificar) y el F1-Score (métrica que balancea la precisión y el recall).

Precisión y Curva ROC

Precisión: Mide la exactitud de las predicciones positivas (de todo lo que el modelo dijo que era positivo, cuántos realmente lo eran).

Curva ROC (Receiver Operating Characteristic): Gráfica que evalúa el punto de balance del clasificador al variar el umbral de decisión, graficando la tasa de verdaderos positivos frente a la tasa de falsos positivos; su área bajo la curva (AUC) resume la calidad general de separación del modelo.

Clasificación Multiclase

Generalización del problema cuando existen más de dos categorías de salida posibles (como clasificar los 10 dígitos distintos del 0 al 9 en MNIST).

Se puede resolver adaptando clasificadores binarios mediante estrategias como One-vs-All (OvA) o utilizando algoritmos que soportan múltiples clases de manera nativa.

Clasificación y Diagnósticos

Marco conceptual donde un sistema toma datos de entrada con incertidumbre, los procesa a través de un modelo de decisión y utiliza herramientas analíticas para diagnosticar en qué falla, cómo se equívoca y qué tan confiable es su comportamiento operativo.
