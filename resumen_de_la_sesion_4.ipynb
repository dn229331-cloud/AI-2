Definición y Valor del Machine Learning Definición de ML: Es un enfoque en el que un programa mejora su rendimiento en una tarea específica a medida que adquiere mayor experiencia a través de los datos (Imagen 4).
Uso e Importancia: El Machine Learning facilita el aprendizaje y el descubrimiento de patrones mediante técnicas de minería de datos (data mining) (Imagen 5).

Ciclo de Desarrollo de ML:

Estudiar el problema (Study the problem).

Entrenar el modelo de ML (Train ML model) utilizando un volumen considerable de datos.

Inspeccionar la solución (Inspect the solution).

Mejorar la comprensión del problema (Better understanding of the problem).

Iterar/Repetir el proceso si es necesario (Repeat if needed).

Tipos de Aprendizaje en Sistemas de ML Las imágenes ilustran distintos enfoques de entrenamiento de algoritmos:
Aprendizaje Supervisado vs. No Supervisado / Clasificación (Imagen 6):

Supervisado (Izquierda): Clasificación basada en etiquetas previas o reglas definidas (contenedores etiquetados con la entrada correcta o incorrecta).

No Supervisado (Derecha): El algoritmo recibe un conjunto de datos heterogéneo y sin etiquetar para identificar agrupaciones, estructuras o patrones por sí mismo.

Aprendizaje por Refuerzo (Reinforcement Learning) (Imagen 7):

Un agente interactúa con un entorno (environment).

El agente observa la situación, selecciona una acción basada en una política (policy) y ejecuta la acción.

Recibe una recompensa o penalización (reward or penalty) según el resultado obtenido.

Con esta retroalimentación, el agente actualiza su política e itera sucesivamente hasta aprender una estrategia óptima.

Estrategias GeneralesLas estrategias de ML definen cómo abordar un problema según la disponibilidad de datos y la complejidad del proyecto:Ingeniería de Características (Feature Engineering): Transformación y selección de las variables más relevantes para facilitar el aprendizaje del modelo antes de entrenarlo.Reducción de Dimensionalidad: Simplificación de conjuntos de datos con muchas variables (evitando la "maldición de la dimensionalidad") mediante técnicas como PCA (Análisis de Componentes Principales).Transfer Learning (Aprendizaje por Transferencia): Reutilización de un modelo preentrenado en una tarea general para adaptarlo a un problema específico con menos datos y tiempo de cómputo.Ensamble de Modelos (Ensemble Methods): Combinación de las predicciones de múltiples modelos (como Random Forest o XGBoost) para obtener un resultado más preciso y robusto que el de un solo modelo.5. Desafíos PrincipalesEl desarrollo de sistemas de ML implica superar barreras técnicas y operativas clave:Cantidad Insuficiente o Mala Calidad de Datos: Datos faltantes, ruina de datos (dirty data), datos sesgados o muestras no representativas del problema real.Sobreajuste (Overfitting): Ocurre cuando el modelo memoriza el conjunto de entrenamiento (incluyendo el ruido) pero falla al generalizar con datos nuevos.Subajuste (Underfitting): El modelo es demasiado simple para capturar las patrones subyacentes de los datos.Deriva del Modelo (Data / Concept Drift): El comportamiento del entorno cambia con el tiempo, haciendo que las predicciones del modelo pierdan precisión si no se actualiza (lo que se mitiga con el Online Learning visto en la Imagen 10).6. Pruebas y ValidaciónGarantizan que el modelo sea confiable antes y después de su despliegue en producción:División de Datos (Data Splitting):Train Set: Para entrenar el modelo.Validation Set: Para ajustar hiperparámetros y comparar modelos.Test Set: Para realizar la evaluación final con datos totalmente no vistos.Validación Cruzada (Cross-Validation): Técnica (como K-Fold) para evaluar el rendimiento del modelo dividiendo iterativamente los datos para reducir el sesgo de selección.Métricas de Evaluación:Clasificación: Precisión (Accuracy), Precision, Recall, F1-Score, Área bajo la curva ROC (AUC-ROC).Regresión: Error Cuadrático Medio (MSE), Error Absoluto Medio (MAE),
.Pruebas A/B y Despliegue en Sombra (Shadow Deployment): Evaluación del comportamiento del modelo con tráfico real comparado contra la solución actual antes de hacer una migración completa.
