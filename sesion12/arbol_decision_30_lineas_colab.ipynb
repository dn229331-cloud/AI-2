# 🌳 Árbol de decisión: ejemplo de la vida diaria
Este Colab construye un árbol **a partir de un pequeño dataset** y explica las 30 líneas de código paso a paso.

```python
1.  import pandas as pd                                      # Importamos pandas para trabajar con tablas.
2.  from sklearn.tree import DecisionTreeClassifier, plot_tree # Importamos el árbol y su función para dibujarlo.
3.  import matplotlib.pyplot as plt                            # Importamos matplotlib para crear gráficos.
4.  datos = pd.DataFrame({                                     # Creamos un dataset pequeño de ejemplo.
5.      'llueve': [1,1,0,0,0,1,0,0],                          # 1 significa que llueve; 0 significa que no.
6.      'lluvia_tarde': [1,0,1,0,0,1,0,1],                    # Indica si esperamos lluvia más tarde.
7.      'mucho_tiempo_fuera': [1,0,1,1,0,0,1,0],             # Indica si estaremos mucho tiempo fuera.
8.      'llevar_paraguas': [1,1,1,0,0,1,0,1]                 # Variable objetivo: 1 = sí, 0 = no.
9.  })                                                       # Cerramos la tabla y guardamos el dataset.
10. X = datos[['llueve','lluvia_tarde','mucho_tiempo_fuera']] # X contiene las características que usará el modelo.
11. y = datos['llevar_paraguas']                             # y contiene la respuesta que queremos predecir.
12. modelo = DecisionTreeClassifier(max_depth=3, random_state=42) # Creamos el árbol y limitamos su profundidad.
13. modelo.fit(X, y)                                         # Entrenamos el árbol usando X e y.
14. print('Profundidad:', modelo.get_depth())                # Mostramos la profundidad que aprendió el árbol.
15. print('Hojas:', modelo.get_n_leaves())                  # Mostramos cuántas decisiones finales tiene.
16. nueva_decision = pd.DataFrame({                          # Creamos un caso nuevo que queremos clasificar.
17.     'llueve': [0],                                       # En el nuevo caso no está lloviendo.
18.     'lluvia_tarde': [1],                                 # Pero se espera lluvia más tarde.
19.     'mucho_tiempo_fuera': [0]                             # No estaremos mucho tiempo fuera.
20. })                                                         # Cerramos los datos del nuevo caso.
21. prediccion = modelo.predict(nueva_decision)[0]            # El árbol predice si debemos llevar paraguas.
22. print('¿Llevar paraguas?', 'Sí' if prediccion == 1 else 'No') # Convertimos 1/0 en una respuesta comprensible.
23. plt.figure(figsize=(14,7))                                # Preparamos el espacio donde dibujaremos el árbol.
24. plot_tree(modelo, feature_names=X.columns, class_names=['No','Sí'], # Dibujamos el árbol con nombres claros.
25.           filled=True, rounded=True, proportion=True)      # Añadimos colores, bordes redondeados y proporciones.
26. plt.title('Árbol de decisión: ¿llevar paraguas?')           # Ponemos un título al gráfico.
27. plt.show()                                                 # Mostramos el árbol en pantalla.
28. print('\nHiperparámetro: max_depth = 3')                    # Indicamos el hiperparámetro que configuramos.
29. print('Una profundidad mayor permite más reglas; una menor hace el árbol más simple.') # Explicamos su efecto.
30. print('Nota: este dataset es didáctico y fue creado manualmente, no son datos reales.') # Aclaramos el origen de los datos.
```

## 🔎 Explicación línea por línea
**1–3:** Importamos las herramientas: `pandas` para datos, `scikit-learn` para Machine Learning y `matplotlib` para visualizar.

**4–9:** Construimos el dataset. Cada fila representa una situación cotidiana y cada columna una característica o la respuesta esperada.

**10–11:** Separamos las **variables de entrada (X)** de la **variable objetivo (y)**. El modelo aprende la relación entre ambas.

**12:** Creamos el árbol. `max_depth=3` es un **hiperparámetro**: limita cuántos niveles puede tener el árbol. `random_state=42` hace reproducible el resultado.

**13:** `fit()` entrena el modelo. Aquí el algoritmo busca divisiones que separen mejor las clases según los datos.

**14–15:** Consultamos dos propiedades del árbol aprendido: profundidad y número de hojas.

**16–20:** Creamos un nuevo ejemplo que no estaba en la tabla original.

**21–22:** Usamos `predict()` para obtener la predicción y transformamos `1/0` en **Sí/No**.

**23–27:** Dibujamos el árbol para poder observar sus preguntas, ramas y hojas.

**28–30:** Relacionamos el resultado con el concepto de **hiperparámetros** y aclaramos que los datos son únicamente didácticos.
