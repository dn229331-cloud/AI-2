# 🌸 Dataset Iris explicado paso a paso

Este cuaderno está pensado para aprender desde cero qué es un **dataset**, cómo está organizado Iris y cómo utilizarlo para Machine Learning.

## ¿Qué es Iris?
Iris es un dataset clásico de Machine Learning que contiene mediciones de flores de tres especies de iris. Cada fila representa una flor y cada columna contiene una característica medida o la especie a la que pertenece.

**Las 4 características:** `sepal length`, `sepal width`, `petal length` y `petal width`.

**La variable objetivo:** especie de iris: *setosa*, *versicolor* o *virginica*.

> Importante: los valores de las medidas están expresados en centímetros.

````python
from sklearn.datasets import load_iris                 # Cargamos el dataset Iris incluido en scikit-learn.
import pandas as pd                                     # Pandas permite trabajar cómodamente con tablas.
import matplotlib.pyplot as plt                         # Matplotlib nos permite crear gráficos.
iris = load_iris()                                      # Cargamos los datos y su información descriptiva.
X = iris.data                                           # X contiene las 4 características de cada flor.
y = iris.target                                         # y contiene la especie codificada como 0, 1 o 2.
print(iris.feature_names)                               # Mostramos los nombres de las cuatro características.
print(iris.target_names)                                # Mostramos los nombres de las tres especies.
df = pd.DataFrame(X, columns=iris.feature_names)        # Convertimos las características en una tabla de pandas.
df['species'] = [iris.target_names[i] for i in y]      # Convertimos 0, 1 y 2 en nombres de especies.
print(df.head())                                       # Observamos las primeras cinco flores del dataset.
print(df.shape)                                        # Mostramos filas y columnas: 150 flores y 5 columnas.
print(df.info())                                       # Revisamos tipos de datos y valores no nulos.
print(df.describe())                                   # Calculamos estadísticas básicas de las características.
print(df['species'].value_counts())                    # Contamos cuántas flores hay de cada especie.
print(df.groupby('species').mean(numeric_only=True))   # Calculamos el promedio de cada medida por especie.
plt.figure(figsize=(8,5))                              # Creamos el espacio para nuestro gráfico.
for especie in iris.target_names:                      # Recorremos las tres especies para dibujarlas.
    datos = df[df['species'] == especie]                # Seleccionamos solamente las flores de esa especie.
    plt.scatter(datos['petal length (cm)'], datos['petal width (cm)'], label=especie) # Dibujamos sus pétalos.
plt.xlabel('Largo del pétalo (cm)')                    # Nombramos el eje horizontal.
plt.ylabel('Ancho del pétalo (cm)')                    # Nombramos el eje vertical.
plt.title('Iris: largo y ancho del pétalo')             # Agregamos un título al gráfico.
plt.legend()                                           # Mostramos la leyenda con las especies.
plt.show()                                             # Mostramos el gráfico.
ejemplo = [[5.1, 3.5, 1.4, 0.2]]                       # Creamos una flor hipotética con cuatro medidas.
print(iris.target_names[iris.target[0]])               # Mostramos la especie real de la primera flor del dataset.
print('Orden de X:', iris.feature_names)               # Recordamos el orden correcto de las características.
print('X = características; y = especie objetivo')      # Resumimos la idea fundamental del aprendizaje supervisado.
````

# 🧠 Explicación profunda

## 1. ¿Qué significa una fila?
Una fila representa **una flor**. Por ejemplo, una fila puede decir que una flor tiene un sépalo de 5.1 cm de largo, 3.5 cm de ancho, un pétalo de 1.4 cm de largo y 0.2 cm de ancho.

## 2. ¿Qué significa una columna?
Una columna representa una **característica (feature)** o la variable objetivo. Las cuatro primeras son mediciones; `species` indica la clase.

## 3. ¿Qué son X e y?
`X` son las pistas que recibe el modelo. `y` es la respuesta que queremos que aprenda a predecir. Por eso Iris es un ejemplo de **aprendizaje supervisado**.

## 4. ¿Por qué aparecen 0, 1 y 2?
Los algoritmos suelen trabajar con valores numéricos. En Iris, las especies están codificadas como:

- `0` → setosa
- `1` → versicolor
- `2` → virginica

## 5. ¿Qué significa `shape = (150, 5)`?
Hay **150 filas** y **5 columnas** en nuestro DataFrame: cuatro características y una columna `species` que agregamos para facilitar la lectura.

## 6. ¿Qué estamos viendo en el gráfico?
Cada punto representa una flor. Al usar largo y ancho del pétalo podemos observar que algunas especies ocupan regiones bastante diferentes del gráfico. Esa separación visual ayuda a entender por qué un algoritmo puede aprender reglas para clasificarlas.

## 7. ¿Cómo se conecta con un árbol de decisión?
Un árbol puede aprender preguntas como: **“¿el ancho del pétalo es menor que cierto valor?”**. Según la respuesta, continúa por una rama distinta. Repite este proceso hasta llegar a una especie predicha.
