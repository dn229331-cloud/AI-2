Resumen General: Introducción al Machine Learning (ML)
1. ¿Qué es el Machine Learning?
Es una disciplina que permite que un programa o sistema mejore su rendimiento en una tarea específica a medida que adquiere mayor experiencia (datos), en lugar de depender únicamente de reglas fijas programadas manualmente.

2. Valor del ML: ¿Por qué usarlo?
Minería de datos: Facilita el aprendizaje automático analizando grandes volúmenes de datos complejos que serían muy difíciles de procesar con la programación tradicional.

Adaptabilidad: Permite descubrir patrones ocultos y resolver problemas dinámicos donde las reglas cambian con el tiempo.

3. Tipos de Sistemas (Niveles de Supervisión)
Dependiendo de cómo se utilicen los datos para entrenar el modelo, existen tres enfoques principales:

Aprendizaje Supervisado: El sistema aprende a partir de datos que ya tienen etiquetas o respuestas conocidas (por ejemplo, clasificar elementos correctos e incorrectos).

Aprendizaje No Supervisado: El modelo analiza datos sin etiquetar para encontrar patrones, estructuras ocultas o agrupar la información por similitudes de forma autónoma.

Aprendizaje por Refuerzo: Un agente aprende a tomar decisiones óptimas interactuando con su entorno, guiado por un sistema de recompensas y castigos para maximizar los aciertos.

4. Flujos de Datos y Ritmos de Aprendizaje
Se refieren a la forma en que los datos fluyen hacia el sistema (ya sea de forma continua o por lotes) y cómo el modelo ajusta su ritmo de aprendizaje para procesar la información de manera eficiente.

5. El Plano de Machine Learning (Ciclo de Vida)
El proceso para desarrollar una solución de ML sigue una serie de pasos iterativos:

Estudiar el problema: Comprender qué se quiere resolver.

Entrenar el modelo de ML: Alimentarlo con "lots of data" (grandes cantidades de datos).

Inspeccionar la solución: Evaluar los resultados obtenidos con herramientas de análisis.

Comprender mejor el problema: Analizar qué se puede mejorar a partir de los resultados.

Repetir si es necesario: Ajustar y volver a iterar para optimizar continuamente el sistema.
