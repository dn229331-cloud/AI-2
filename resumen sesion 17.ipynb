Retos de los gradientes  
(Desvanecimiento y Explosión)El problema raíz:  
Al propagar el error hacia atrás mediante la regla de la cadena (Backpropagation), las derivadas se multiplican capa por capa. Si son menores a 1, el gradiente se desvanece (las capas iniciales no aprenden); si son mayores a 1, el gradiente explota (el modelo se vuelve numéricamente inestable o diverge).El cuello de botella de la sigmoide:  
Las funciones tradicionales como la sigmoide o la tangente hiperbólica comprimen las entradas en un rango estricto ($[0, 1]$ o $[-1, 1]$).  

En sus extremos, sus derivadas se acercan peligrosamente a cero, matando por completo el flujo de información.  

2. Inicialización de pesosControl de la varianza: Para que la señal fluya correctamente por decenas de capas, la varianza de las activaciones debe mantenerse constante entre la entrada y la salida de cada capa.Métodos estándar (corrección de "molos"):Glorot (o Xavier): Diseñada para funciones simétricas como tanh o sigmoide, ajusta la varianza según el número de neuronas de entrada y salida.He (Kaiming): Específicamente optimizada para ReLU y sus variantes, duplicando la varianza para compensar las neuronas que se "apagan" a cero.  

3. Funciones de activaciónLa revolución de ReLU: Rompió el paradigma de que las activaciones debían imitar neuronas biológicas acotadas. Al no saturarse para valores positivos ($\max(0, x)$), permite que los gradientes fluyan sin atenuarse.Evolución hacia funciones suaves:ELU y SELU: Introducen un componente exponencial para valores negativos, acercando la media de las activaciones a cero y logrando auto-normalización (especialmente SELU), lo que previene la muerte de neuronas.GELU, Swish y Mish: Son funciones suaves, continuas y no monótonas que permiten un flujo de gradiente más fluido. GELU es el estándar de oro en arquitecturas de vanguardia como los Transformers (GPT, BERT), mientras que Swish/Mish brillan en visión por computador.  

4. Normalización (Normalización por lotes / Batch Normalization)El mecanismo: Actúa como una capa que centra (media cero), escala (desviación estándar uno) y desplaza/re-escala (usando parámetros aprendibles $\gamma$ y $\beta$) las activaciones de un lote antes de la siguiente función de activación.Por qué es un "superpoder": Mitiga el cambio de covarianza interno (internal covariate shift), estabiliza el entrenamiento, actúa como un regularizador ligero (reduciendo el sobreajuste) y permite utilizar tasas de aprendizaje (learning rates) mucho más agresivas.  

5. Aprendizaje por transferencia (Transfer Learning)El concepto: Reutilizar el conocimiento que una red ya adquirió resolviendo un problema masivo para aplicarlo a una tarea nueva con pocos datos.Estrategia de despliegue:Congelar (freeze) las capas base (que extraen patrones genéricos como bordes, texturas o sintaxis básica).Entrenar únicamente las capas superiores (fine-tuning) con la nueva tarea.Opcionalmente, descongelar progresivamente (gradual unfreezing) las capas inferiores con una tasa de aprendizaje muy baja para adaptarlas sutilmente al nuevo dominio.Nota: Esta técnica, junto con el preentrenamiento, impulsó el renacimiento masivo de las redes neuronales en la última década.  

6. Preentrenamiento (Pre-training)El concepto enriquecido: Es la fase inicial donde un modelo aprende representaciones universales del mundo a partir de una gran cantidad de datos (frecuentemente mediante aprendizaje no supervisado o auto-supervisado, como predecir palabras ocultas en texto o parches en imágenes).Propósito: En lugar de entrenar una red desde cero (scratch) con pesos aleatorios, el preentrenamiento le otorga al modelo una "intuición" o comprensión previa del lenguaje, la visión o el dominio, reduciendo drásticamente el tiempo de entrenamiento y los datos necesarios para tareas específicas posteriores (downstream tasks).
