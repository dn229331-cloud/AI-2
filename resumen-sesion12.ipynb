​01. Entrenando Árboles  

​¿Qué es? Es el proceso mediante el cual el algoritmo aprende de los datos históricos para construir la estructura de decisiones (nodos y ramas).  
El objetivo es encontrar las preguntas (divisiones o splits) que mejor separan las clases o los valores objetivo.  

​Cómo funciona: El algoritmo evalúa todas las variables disponibles y busca el "punto de corte" óptimo que divida los datos de manera que los grupos resultantes sean lo más homogéneos (puros) posibles.  

​Ejemplo práctico: Imagina que quieres predecir si un cliente va a comprar una motocicleta. El árbol analiza tu base de datos y decide que la primera gran pregunta (raíz) sea: ¿El cliente tiene licencia de conducción? Si es "Sí", evalúa una segunda pregunta: ¿Su salario es mayor a cierto monto? El árbol se construye ramificándose hasta tomar una decisión final (Comprará / No comprará).  

​02. Predicciones y Probabilidades  

​¿Qué es? La capacidad del árbol entrenado para evaluar un nuevo caso y arrojar un resultado, el cual no solo incluye la categoría final, sino también la confianza o probabilidad de que esa predicción sea correcta.  

​Cómo funciona: Un nuevo dato recorre el árbol desde la raíz respondiendo a las preguntas de cada nodo hasta llegar a una "hoja" final. La predicción se basa en la mayoría de los casos de entrenamiento que cayeron en esa misma hoja.  

​Ejemplo práctico: Si un nuevo cliente llega al sistema y el árbol lo ubica en una hoja final donde de 100 personas históricas similares, 80 compraron y 20 no, el modelo te dará dos cosas:
​Predicción: Comprará.
​Probabilidad (Proba): 80% de confianza en esa predicción.  

​03. CART e Impureza  

​¿Qué es? CART (Classification and Regression Trees) es el algoritmo estándar para construir estos árboles. La impureza mide qué tan mezclados están los datos en un nodo.  

​Conceptos clave de impureza:
​Un nodo es puro (impureza 0) si todos los elementos pertenecen a una sola categoría.
​Se usan métricas matemáticas como el Índice Gini o la Entropía (Ganancia de Información) para medir cuánt desorden hay. El algoritmo siempre elige la división que reduzca al máximo esta impureza.  

​Ejemplo práctico: Si en un nodo tienes 50 compradores y 50 no compradores, la impureza es máxima (está 50/50). Si aplicas un filtro (ej. edad < 30) y logras separar el grupo en 45 compradores y 5 no compradores, la impureza bajó drásticamente porque el nodo se volvió mucho más puro.  

​04. Regularización  

​¿Qué es? Conjunto de restricciones que se le aplican al árbol para evitar el sobreajuste (overfitting). Si dejas que un árbol crezca libremente, memorizará tanto los datos de entrenamiento que funcionará mal con datos nuevos.  

​Técnicas comunes:  

​Profundidad máxima (max_depth): Limitar cuántos niveles de preguntas puede tener el árbol.
​Mínimo de muestras por hoja (min_samples_leaf): Exigir que una hoja final tenga un número mínimo de casos para no crear reglas demasiado específicas o basadas en muy pocos datos.  

​Ejemplo práctico: Sin regularización, un árbol podría crear una regla absurda como: "Si el cliente tiene exactamente 27 años, mide 1.74m y compró un martes, entonces no comprará". Eso es memorizar ruido. La regularización frena el crecimiento para mantener reglas lógicas y generales.  

​05. Árboles de Regresión  

​¿Qué es? Una variante de los árboles de decisión que no sirve para clasificar en categorías (como Sí/No o A/B/C), sino para predecir un número continuo.
​Cómo funciona: En lugar de medir impureza con Gini o Entropía, se utiliza la Varianza o el Error Cuadrático Medio (MSE).   
Las hojas finales del árbol devuelven un valor numérico que representa el promedio de los datos que cayeron en esa hoja.  

​Ejemplo práctico: Predecir el precio de venta de una motocicleta usada. El árbol evalúa el cilindraje, el año y el kilometraje, y en lugar de decir "Es cara o barata", te da un valor numérico exacto: "El precio estimado es de $4,500,000 COP" (basado en el promedio de las motos con características similares en las hojas finales).  

​06. Limitaciones  

​¿Qué es? Desventajas y puntos débiles inherentes a los árboles de decisión que debes tener en cuenta al usarlos en proyectos reales.
​Principales limitaciones:
​Inestabilidad (Alta varianza): Un pequeño cambio en los datos de entrenamiento (añadir o quitar un par de registros) puede hacer que toda la estructura del árbol cambie radicalmente.  

​Tienden al sobreajuste: Si no se podan o regulan adecuadamente, pierden capacidad de generalización.
​Fronteras de decisión ortogonales: Los árboles solo pueden hacer cortes rectos y verticales/horizontales en los datos, lo que a veces complica modelar relaciones muy suaves o diagonales. (Nota: Esto se suele solucionar combinando muchos árboles, dando origen a los modelos de Ensamble como Random Forest
