Cracion de chatbot

Concepto general: Desarrollo e implementación de un asistente virtual conversacional automatizado diseñado específicamente para el sector de alimentos y bebidas (panadería), con el propósito de optimizar la atención al cliente, exhibir el catálogo de productos y facilitar la toma de pedidos en línea de manera interactiva.

Propósito del sistema: Automatizar el flujo de atención inicial mediante una interfaz amigable guiada por una personalidad virtual ("Elena"), reduciendo tiempos de respuesta y ofreciendo a los usuarios acceso inmediato a información clave como menús, datos de contacto, horarios y ubicaciones comerciales.
