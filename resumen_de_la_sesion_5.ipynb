# resumen de la sesion 5

El proceso KDD (Knowledge Discovery in Databases o Descubrimiento de Conocimiento en Bases de Datos) transforma grandes volúmenes de datos desorganizados en información útil y aplicable.
​¿Qué es el Proceso KDD?
Es una metodología estructurada que convierte el "caos" de los datos brutos en conocimiento valioso, similar a extraer un diamante de una mina de rocas.
​Evolución Histórica
​1989: Usama Fayyad y Piatetsky-Shapiro acuñan el término KDD en el primer taller oficial.
​1996: Fayyad y su equipo formalizan los pasos detallados del proceso paso a paso.
​2016: Se publica la expansión del marco adaptado para entornos de Big Data y la nube.
​Los 5 Pasos del KDD
​1. Selección: Aislar y extraer el subconjunto de datos correcto y relevante para el objetivo.
​2. Preproceso: Limpiar los datos anómalos, eliminar ruido y corregir errores o inconsistencias.
​3. Transformación: Estructurar y formatear los datos matemáticamente para que sean aptos para el análisis.
​4. Minería de Datos (Data Mining): Aplicar algoritmos especializados para buscar patrones, tendencias o correlaciones ocultas.
​5. Evaluación: Interpretar los resultados obtenidos para verificar si realmente representan conocimiento útil y válido.
​Aplicaciones Reales por Industria
​Finanzas: Detección de fraudes y transacciones anómalas.
​Salud: Diagnóstico temprano de enfermedades a partir del historial médico.
​Retail (Comercio): Análisis de canasta de compras para predecir comportamiento del consumidor.
​Consideraciones Éticas
El uso avanzado de reconocimiento de patrones y minería de datos plantea importantes retos sobre la privacidad de los usuarios y el manejo de su información personal.
