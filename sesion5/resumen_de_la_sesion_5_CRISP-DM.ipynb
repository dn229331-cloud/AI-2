# resumen de la sesion  5 CRISP-DM

El CRISP-DM (Cross-Industry Standard Process for Data Mining) es el plano maestro estándar inter-industrias diseñado para guiar proyectos de minería de datos y ciencia de datos de forma estructurada.
​Orígenes e Historia de CRISP-DM
​1996: Concebido originalmente por Daimler-Chrysler, SPSS y NCR como un enfoque unificado para proyectos de datos.
​1999: Se publica de manera oficial la versión CRISP-DM 1.0 consolidándose como un estándar inter-industrias.
​2015: IBM introduce una extensión del modelo denominada ASUM-DM para adaptarlo a entornos modernos.
​Las 5 Fases del Modelo CRISP-DM
​01. Orígenes y Práctica: Introducción a la metodología y fundamentos del ciclo de vida de los proyectos basados en datos.
​02. Negocio y Datos: Comprensión inicial de los objetivos del negocio y la recopilación de los datos necesarios.
​03. Preparación de Datos: Limpieza, transformación y acondicionamiento del conjunto de datos para su posterior análisis.
​04. Modelado y Evaluación: Aplicación de técnicas de modelado matemático o estadístico y evaluación rigurosa de los resultados obtenidos.
​05. Despliegue y Futuro: Implementación de los modelos en entornos de producción y seguimiento de su evolución a futuro.
