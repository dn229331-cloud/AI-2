# 🏠 Predicción de precios de casas

## Objetivo

Desarrollar un modelo de Machine Learning capaz de predecir el precio medio de una vivienda utilizando el **California Housing Dataset** de Scikit-Learn.

El proyecto incluye exploración de datos, preparación, entrenamiento de un modelo de Regresión Lineal, evaluación y una predicción para una vivienda hipotética.

## 1. Importación de librerías

En esta sección importamos las herramientas necesarias para manipular datos, realizar cálculos, crear gráficos y construir el modelo de Machine Learning.

```python
# Importamos pandas para trabajar con los datos en forma de tablas.
import pandas as pd

# Importamos numpy para realizar operaciones matemáticas y numéricas.
import numpy as np

# Importamos matplotlib para crear gráficos.
import matplotlib.pyplot as plt

# Importamos seaborn para crear visualizaciones estadísticas.
import seaborn as sns

# Importamos el dataset California Housing desde Scikit-Learn.
from sklearn.datasets import fetch_california_housing

# Importamos la función para dividir los datos en entrenamiento y prueba.
from sklearn.model_selection import train_test_split

# Importamos el modelo de Regresión Lineal.
from sklearn.linear_model import LinearRegression

# Importamos las métricas que utilizaremos para evaluar el modelo.
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

## 2. Carga del dataset

Utilizamos el California Housing Dataset, un conjunto de datos estándar para problemas de regresión. La variable objetivo `MedHouseVal` representa el valor medio de las viviendas en unidades de 100.000 dólares.

```python
# Cargamos el California Housing Dataset y solicitamos que también se entregue como DataFrame.
housing = fetch_california_housing(as_frame=True)

# Guardamos el DataFrame completo en la variable df.
df = housing.frame

# Mostramos las primeras cinco filas para conocer la estructura de los datos.
df.head()
```

## 3. Exploración de los datos

Antes de entrenar el modelo debemos conocer el tamaño del dataset, sus columnas, tipos de datos y estadísticas principales.

```python
# Mostramos la cantidad de filas y columnas del dataset.
print("Dimensiones del dataset:", df.shape)

# Mostramos información sobre las columnas, tipos de datos y valores no nulos.
df.info()

# Mostramos estadísticas descriptivas de las variables numéricas.
df.describe()
```

```python
# Contamos los valores nulos existentes en cada columna.
print("Valores nulos por columna:")
print(df.isnull().sum())
```

## 4. Variables del dataset

| Variable | Descripción |
|---|---|
| `MedInc` | Ingreso medio de los hogares |
| `HouseAge` | Edad media de las viviendas |
| `AveRooms` | Número promedio de habitaciones |
| `AveBedrms` | Número promedio de dormitorios |
| `Population` | Población del área |
| `AveOccup` | Ocupación promedio por vivienda |
| `Latitude` | Latitud |
| `Longitude` | Longitud |
| `MedHouseVal` | Valor medio de las viviendas; es la variable que queremos predecir |

## 5. Visualización de la variable objetivo

```python
# Creamos una figura con un tamaño de 10 por 6 pulgadas.
plt.figure(figsize=(10, 6))

# Creamos un histograma para observar la distribución de los precios.
sns.histplot(df["MedHouseVal"], bins=50, kde=True)

# Añadimos un título al gráfico.
plt.title("Distribución de los precios de las viviendas")

# Añadimos el nombre del eje X.
plt.xlabel("Precio de la vivienda (unidades de $100.000)")

# Añadimos el nombre del eje Y.
plt.ylabel("Cantidad de viviendas")

# Mostramos el gráfico.
plt.show()
```

## 6. Separación de variables predictoras y variable objetivo

`X` contiene las características que utilizaremos para hacer las predicciones.

`y` contiene el precio que queremos predecir.

```python
# Eliminamos la columna MedHouseVal de X porque es justamente el valor que queremos predecir.
X = df.drop("MedHouseVal", axis=1)

# Guardamos MedHouseVal como nuestra variable objetivo.
y = df["MedHouseVal"]

# Mostramos las dimensiones de las variables para comprobar la separación.
print("Dimensiones de X:", X.shape)
print("Dimensiones de y:", y.shape)
```

## 7. División de los datos

Separamos el 80 % de los datos para entrenamiento y el 20 % para prueba. El modelo aprende con el conjunto de entrenamiento y posteriormente se evalúa con datos que no utilizó durante el entrenamiento.

```python
# Dividimos X e y en datos de entrenamiento y datos de prueba.
# test_size=0.20 significa que el 20 % se utilizará para evaluar el modelo.
# random_state=42 permite obtener la misma división cada vez que ejecutamos el código.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Mostramos el tamaño de cada conjunto.
print("Datos de entrenamiento:", X_train.shape)
print("Datos de prueba:", X_test.shape)
```

## 8. Creación del modelo

Utilizaremos **Regresión Lineal**, un algoritmo de aprendizaje supervisado que busca establecer una relación entre las características de las viviendas y su precio.

```python
# Creamos una instancia del modelo de Regresión Lineal.
modelo = LinearRegression()
```

## 9. Entrenamiento del modelo

```python
# Entrenamos el modelo utilizando las características y precios del conjunto de entrenamiento.
modelo.fit(X_train, y_train)

# Mostramos un mensaje para confirmar que el entrenamiento terminó.
print("El modelo fue entrenado correctamente.")
```

## 10. Predicciones

Ahora utilizamos el modelo entrenado para estimar los precios correspondientes a las viviendas del conjunto de prueba.

```python
# Utilizamos el modelo entrenado para generar predicciones sobre los datos de prueba.
y_pred = modelo.predict(X_test)

# Mostramos las primeras diez predicciones.
print("Primeras 10 predicciones:")
print(y_pred[:10])
```

## 11. Evaluación del modelo

Utilizaremos MAE, MSE, RMSE y R² para medir qué tan bien funciona el modelo.

```python
# Calculamos el Error Absoluto Medio (MAE).
mae = mean_absolute_error(y_test, y_pred)

# Calculamos el Error Cuadrático Medio (MSE).
mse = mean_squared_error(y_test, y_pred)

# Calculamos la raíz del Error Cuadrático Medio (RMSE).
rmse = np.sqrt(mse)

# Calculamos el coeficiente de determinación R².
r2 = r2_score(y_test, y_pred)

# Mostramos las métricas obtenidas.
print(f"MAE: {mae:.4f}")
print(f"MSE: {mse:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"R²: {r2:.4f}")
```

### ¿Cómo interpretar las métricas?

- **MAE:** indica el error absoluto promedio de las predicciones.
- **MSE:** calcula el promedio de los errores al cuadrado y penaliza más los errores grandes.
- **RMSE:** es la raíz del MSE y permite interpretar el error en la misma escala de la variable objetivo.
- **R²:** indica qué proporción de la variabilidad de los precios es explicada por el modelo. Un valor más cercano a 1 representa un mejor ajuste.

## 12. Comparación entre precios reales y predichos

```python
# Creamos una figura para comparar los valores reales con las predicciones.
plt.figure(figsize=(10, 6))

# Creamos un gráfico de dispersión con los precios reales en X y los predichos en Y.
plt.scatter(y_test, y_pred, alpha=0.5)

# Añadimos un título al gráfico.
plt.title("Precios reales vs precios predichos")

# Etiquetamos el eje X.
plt.xlabel("Precio real")

# Etiquetamos el eje Y.
plt.ylabel("Precio predicho")

# Mostramos el gráfico.
plt.show()
```

## 13. Predicción para una vivienda hipotética

Para demostrar el funcionamiento del modelo, creamos una vivienda con características hipotéticas y calculamos su precio estimado.

```python
# Creamos un DataFrame con las características de una vivienda hipotética.
nueva_casa = pd.DataFrame({
    "MedInc": [5.0],
    "HouseAge": [20],
    "AveRooms": [6],
    "AveBedrms": [2],
    "Population": [1000],
    "AveOccup": [3],
    "Latitude": [34.0],
    "Longitude": [-118.0]
})

# Utilizamos el modelo entrenado para predecir el valor de la vivienda.
prediccion = modelo.predict(nueva_casa)

# Convertimos el resultado de unidades de $100.000 a dólares.
precio_dolares = prediccion[0] * 100000

# Mostramos el precio estimado.
print(f"Precio estimado de la vivienda: ${precio_dolares:,.2f}")
```
