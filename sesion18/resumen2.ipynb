Trabajo Técnico de Ingeniería de Software: Arquitectura y Diseño de Sistemas de Visión por ComputadoraContexto de la Investigación: Análisis avanzado de los modelos de aprendizaje profundo aplicados al procesamiento de imágenes, basado en la estructuración modular de sistemas de visión artificial.   Nivel Académico: Ingeniería de Software / Inteligencia Artificial Aplicada.Introducción y Marco TeóricoLa Visión por Computadora (Computer Vision) representa una de las ramas más exigentes de la ingeniería de software moderna. Requiere la integración de modelos matemáticos complejos, procesamiento masivo en paralelo (GPU/TPU) y tuberías de datos (pipelines) altamente optimizadas. A partir de la estructura temática observada en las diapositivas de referencia, se expone a continuación el desarrollo detallado de los seis bloques fundamentales bajo un enfoque de ingeniería de software.01. Inspiración BiológicaFundamento Arquitectónico: Las redes neuronales convolucionales (CNN) emulan la organización del sistema visual biológico, específicamente el funcionamiento de la corteza visual primaria descubierta por Hubel y Wiesel.Implicaciones de Diseño: Se modela el flujo de información imitando la recepción de estímulos a través de campos receptivos locales, permitiendo que las capas iniciales detecten patrones primitivos (bordes, esquinas) y las capas profundas reconozcan semánticas complejas (objetos, rostros).02. Capas y Filtros (Convoluciones)Operación Matemática y de Software: La convolución discreta aplica un núcleo (kernel) de pesos entrenables $K$ sobre una matriz de entrada $I$ (imagen o mapa de características):$$S(i,j) = (I * K)(i,j) = \sum_m \sum_n I(i-m, j-n) K(m,n)$$Patrones de Diseño: Aplicación de hiperparámetros clave como el Stride (zancada) para el submuestreo espacial y el Padding (relleno) para conservar las dimensiones espaciales en los bordes de la matriz.03. Agrupación (Pooling)Reducción de Dimensionalidad: Mecanismo de abstracción diseñado para disminuir la carga computacional y mitigar el sobreajuste (overfitting).Tipos de Capas:Max Pooling: Extrae el valor máximo de una región vecina, preservando las características más prominentes.Average Pooling: Calcula el promedio matemático del vecindario, suavizando la representación de la imagen.04. Arquitecturas FamosasEvolución de Modelos Base:LeNet-5: Precursora de las CNN modernas orientada al reconocimiento de dígitos manuscritos.AlexNet / VGGNet: Demostraron el impacto del uso de múltiples capas convolucionales apiladas y el hardware de GPU.ResNet (Residual Networks): Introdujo las conexiones de salto (skip connections), resolviendo el problema del desvanecimiento del gradiente (vanishing gradient) en redes ultraprofundas.05. Modelos Preentrenados y Transfer LearningOptimización de Recursos: En la ingeniería de software actual, entrenar redes desde cero es ineficiente en términos de tiempo y coste computacional.Estrategia de Transferencia: Se reutilizan pesos sinápticos generados en datasets masivos (como ImageNet). Mediante la técnica de Fine-Tuning, se congelan las capas iniciales de extracción de características y se reentrenan únicamente las capas densas finales para adaptarlas a un dominio de negocio específico.06. Detección de Objetos y Despliegue en ProducciónCapacidades Avanzadas: A diferencia de la clasificación simple, la detección de objetos combina la localización espacial (delimitación mediante Bounding Boxes) y la categorización simultánea.Integración de Software: Implementación de frameworks como YOLO (You Only Look Once) o SSD (Single Shot MultiBox Detector) encapsulados en microservicios mediante contenedores Docker, listos para realizar inferencia en tiempo real en entornos de producción.
